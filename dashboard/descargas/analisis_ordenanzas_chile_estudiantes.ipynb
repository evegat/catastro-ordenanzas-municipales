```python
# 1. Carga de librerías esenciales para análisis de políticas públicas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
print('Entorno listo para investigación de gobernanza local.')
```

## 📊 2. Carga y Exploración del Dataset Nacional
Cargamos el archivo consolidado `catastro_ordenanzas_nacional_2026.csv`.

```python
# Cargar dataset oficial
url_dataset = 'catastro_ordenanzas_nacional_2026.csv'
df = pd.read_csv(url_dataset)

print(f'Total de registros normativos: {len(df)}')
print(f'Comunas con datos observados: {df["comuna"].nunique()}')
df.head(5)
```

## 🏷️ 3. Distribución por Eje Temático (Materias)
¿Cuáles son las materias más reguladas por los municipios chilenos?

```python
materias_conteo = df['materia'].value_counts()
print(materias_conteo)

plt.figure(figsize=(10, 5))
sns.barplot(x=materias_conteo.values, y=materias_conteo.index, palette='viridis')
plt.title('Distribución Nacional de Ordenanzas Municipales por Materia', fontsize=14, fontweight='bold')
plt.xlabel('Número de Ordenanzas')
plt.ylabel('Eje Temático')
plt.tight_layout()
plt.show()
```

## 🗺️ 4. Análisis Territorial Comparado por Región
¿Existe concentración de ordenanzas en la Región Metropolitana o hay actividad normativa descentralizada?

```python
regiones_conteo = df['region_nombre'].value_counts()

plt.figure(figsize=(12, 6))
sns.barplot(x=regiones_conteo.values, y=regiones_conteo.index, palette='magma')
plt.title('Concentración Regional de Ordenanzas Municipales en Chile', fontsize=14, fontweight='bold')
plt.xlabel('Total de Normas Registradas')
plt.ylabel('Región')
plt.tight_layout()
plt.show()
```
